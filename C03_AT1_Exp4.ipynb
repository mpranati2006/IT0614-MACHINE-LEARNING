```python
# Install pgmpy (Run once in Google Colab)
!pip install pgmpy -q

from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination

# Create Bayesian Belief Network
model = DiscreteBayesianNetwork([
    ('Flu', 'Fever'),
    ('Flu', 'Cough'),
    ('Flu', 'BodyPain')
])

# Prior Probability of Flu
cpd_flu = TabularCPD(
    variable='Flu',
    variable_card=2,
    values=[[0.8], [0.2]]
)

# Fever given Flu
cpd_fever = TabularCPD(
    variable='Fever',
    variable_card=2,
    values=[
        [0.8, 0.1],   # Fever = No
        [0.2, 0.9]    # Fever = Yes
    ],
    evidence=['Flu'],
    evidence_card=[2]
)

# Cough given Flu
cpd_cough = TabularCPD(
    variable='Cough',
    variable_card=2,
    values=[
        [0.7, 0.2],   # Cough = No
        [0.3, 0.8]    # Cough = Yes
    ],
    evidence=['Flu'],
    evidence_card=[2]
)

# Body Pain given Flu
cpd_body = TabularCPD(
    variable='BodyPain',
    variable_card=2,
    values=[
        [0.9, 0.3],   # BodyPain = No
        [0.1, 0.7]    # BodyPain = Yes
    ],
    evidence=['Flu'],
    evidence_card=[2]
)

# Add CPDs
model.add_cpds(cpd_flu, cpd_fever, cpd_cough, cpd_body)

# Check Model
print("Model Valid:", model.check_model())

# Inference
infer = VariableElimination(model)

print("\nCase 1: Fever = Yes")
print(infer.query(['Flu'], evidence={'Fever':1}))

print("\nCase 2: Fever = Yes, Cough = Yes")
print(infer.query(['Flu'], evidence={'Fever':1, 'Cough':1}))

print("\nCase 3: Fever = Yes, Cough = Yes, BodyPain = Yes")
print(infer.query(['Flu'], evidence={'Fever':1, 'Cough':1, 'BodyPain':1}))

print("\nCase 4: Fever = No")
print(infer.query(['Flu'], evidence={'Fever':0}))

# Modify Conditional Probability
print("\nAfter Changing Fever Probability")

cpd_fever_new = TabularCPD(
    variable='Fever',
    variable_card=2,
    values=[
        [0.8, 0.3],
        [0.2, 0.7]
    ],
    evidence=['Flu'],
    evidence_card=[2]
)

model.remove_cpds(cpd_fever)
model.add_cpds(cpd_fever_new)

infer = VariableElimination(model)

print(infer.query(['Flu'], evidence={'Fever':1}))
```
