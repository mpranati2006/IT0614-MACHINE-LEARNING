```python
# K-Nearest Neighbour (KNN) Algorithm

import math

# Small dataset
X = [
    [1, 2],
    [2, 3],
    [3, 1],
    [5, 4],
    [6, 5],
    [7, 3]
]

# Class labels
y = ['A', 'A', 'A', 'B', 'B', 'B']

# Test data
test = [4, 3]

# Take K value from user
k = int(input("Enter the value of K: "))

# Calculate Euclidean distance
distances = []

for i in range(len(X)):
    distance = math.sqrt(
        (X[i][0] - test[0])**2 +
        (X[i][1] - test[1])**2
    )
    distances.append((distance, y[i]))

# Sort distances
distances.sort()

# Select K nearest neighbours
nearest = distances[:k]

print("\nNearest Neighbours:")
for d, label in nearest:
    print("Distance =", round(d, 2), "Class =", label)

# Count class labels
votes = {}

for d, label in nearest:
    votes[label] = votes.get(label, 0) + 1

# Predicted class
predicted_class = max(votes, key=votes.get)

print("\nTest Data:", test)
print("Predicted Class:", predicted_class)

# Calculate training accuracy
correct = 0

for i in range(len(X)):
    test_point = X[i]

    other_points = X[:i] + X[i+1:]
    other_labels = y[:i] + y[i+1:]

    temp_distances = []

    for j in range(len(other_points)):
        distance = math.sqrt(
            (other_points[j][0] - test_point[0])**2 +
            (other_points[j][1] - test_point[1])**2
        )
        temp_distances.append((distance, other_labels[j]))

    temp_distances.sort()
    nearest_neighbors = temp_distances[:k]

    votes = {}
    for d, label in nearest_neighbors:
        votes[label] = votes.get(label, 0) + 1

    prediction = max(votes, key=votes.get)

    if prediction == y[i]:
        correct += 1

accuracy = (correct / len(X)) * 100

print("Accuracy:", round(accuracy, 2), "%")
```
